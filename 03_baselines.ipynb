# 03 · Baselines

O "piso" de comparação. Se uma rede não supera esses métodos, o erro baixo dela é mérito da série, não da rede.

| Baseline | Previsão | Por que importa |
|---|---|---|
| `naive` | repete o último valor | atalho mais simples possível |
| `sazonal` | repete o último dia (24 passos no ETTh1, 144 no Weather) | captura o ciclo diário sem aprender nada |
| `media_janela` | média da janela de entrada | é exatamente o que um modelo com RevIN prevê se a rede devolver zero |

`naive` e `sazonal` não dependem da janela. Eles são avaliados em cada L só para facilitar as tabelas, e o resultado é idêntico porque os alvos são os mesmos.
